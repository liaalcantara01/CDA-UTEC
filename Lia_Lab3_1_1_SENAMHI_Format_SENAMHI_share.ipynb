```python
# Ejercicio: Estación

# Leer los datos de las estaciones y crear un DataFrame combinado
import os

ruta = '/content/drive/MyDrive/Senamhi'

files = os.listdir(ruta)

files
```

```python
# Parsear el nombre del archivo para obtener el código y nombre de la estación

station = 'qc00151209_chosica.txt'

codigo = station.split('_')[0].replace('qc', '')
nombre = station.split('_')[1].replace('.txt', '')

print(f"Código: {codigo}, Nombre: {nombre}")
```

```python
# Leer el archivo CSV de la estación seleccionada
import pandas as pd

station_df = pd.read_table(
    f'{ruta}/{station}',
    header=None,
    sep=r'\s+',
    na_values=['-999', '-99.9', '-3256']
)

station_df.columns = ['Año', 'Mes', 'Dia', 'Pr', 'Tmax', 'Tmin']  # Asignar nombres de columnas

# Formatear las columnas
station_df['Año'] = station_df['Año'].astype(int)
station_df['Mes'] = station_df['Mes'].astype(int)
station_df['Dia'] = station_df['Dia'].astype(int)

station_df
```

```python
# Precipitacion mensual
precipitacion_mensual = station_df.groupby(['Año', 'Mes'])['Pr'].mean()

# Precipitacion anual
precipitacion_anual = station_df.groupby('Año')['Pr'].mean()

# Dia mas lluvioso
dia_mas_lluvioso = station_df.loc[station_df['Pr'].idxmax()]


print(f"Precipitación mensual:\n{precipitacion_mensual}\n")
print(f"Precipitación anual:\n{precipitacion_anual}\n")
print(f"El día más lluvioso fue el {dia_mas_lluvioso['Año']:.0f}-{dia_mas_lluvioso['Mes']:.0f}-{dia_mas_lluvioso['Dia']:.0f} con {dia_mas_lluvioso['Pr']} mm de precipitación.")
```

```python
# Exportar los resultados

import os

ruta_resultados = '/content/drive/MyDrive/Resultados_SENAMHI'

# Creo una carpeta para gusradar los resultados
os.makedirs(ruta_resultados, exist_ok=True)

precipitacion_mensual.to_csv(
    f'{ruta_resultados}/precipitacion_mensual_chosica.csv'
)

precipitacion_anual.to_csv(
    f'{ruta_resultados}/precipitacion_anual_chosica.csv'
)

dia_mas_lluvioso.to_frame().T.to_csv(
    f'{ruta_resultados}/dia_mas_lluvioso_chosica.csv',
    index=False
)

print("Resultados exportados correctamente.")
```

### Ejercicio 2: Datos climáticos multiestación

Crea un nuevo DataFrame con datos de 5 estaciones SENAMHI de diferentes regiones del Perú (https://www.senamhi.gob.pe/site/descarga-datos/).

Luego:
1. Paresear el nombres de la estaciones y leer los archivo.
1. Unir las db (Fecha, Est1, Est2, ...)
1. Calcula el promedio mensual de precipitación por región
2. Identifica el mes más lluvioso en cada estación
4. Exporta el resultado formateado a CSV

```python
# Primero leer y unir las 5 estaciones

import pandas as pd
import os

archivos = [archivo for archivo in os.listdir(ruta) if archivo.endswith('.txt')]

estaciones = []

for archivo in archivos:

    # Nombre de la estación:
    nombre = archivo.replace('.txt', '').split('_', 1)[1]

    # Leer los datos:
    df = pd.read_table(
        f'{ruta}/{archivo}',
        header=None,
        sep=r'\s+',
        na_values=['-999', '-99.9', '-3256']
    )

    df.columns = ['Año', 'Mes', 'Dia', 'Pr', 'Tmax', 'Tmin']

    # Crear fecha:
    df['Fecha'] = pd.to_datetime(
        dict(year=df['Año'], month=df['Mes'], day=df['Dia'])
    )

    # Guardamos solo fecha y precipitación:
    estaciones.append(
        df[['Fecha', 'Pr']].rename(columns={'Pr': nombre}).set_index('Fecha')
    )

# Unimos todas las estaciones:
multiestacion_df = pd.concat(estaciones, axis=1).reset_index()

multiestacion_df
```

```python
# Promedio mensual de precipitación

precipitacion_mensual_estaciones = (
    multiestacion_df
    .set_index('Fecha')
    .resample('ME')
    .mean()
)

# Mes más lluvioso de cada estación

resultado_mes_lluvioso = pd.DataFrame({
    'Mes_mas_lluvioso': precipitacion_mensual_estaciones.idxmax(),
    'Precipitacion_promedio_mm': precipitacion_mensual_estaciones.max()
})

# Para la fecha mostramos solo año y mes
resultado_mes_lluvioso['Mes_mas_lluvioso'] = (
    resultado_mes_lluvioso['Mes_mas_lluvioso'].dt.strftime('%Y-%m')
)

print("PROMEDIO MENSUAL:")
display(precipitacion_mensual_estaciones)

print("\nMES MÁS LLUVIOSO POR ESTACIÓN:")
display(resultado_mes_lluvioso)
```

```python
# Exportar los resultados del Ejercicio 2 en la misma carpeta que se creo para el ejercicio 1

multiestacion_df.to_csv(
    f'{ruta_resultados}/datos_multiestacion.csv',
    index=False,
    sep=';',
    encoding='utf-8-sig'
)

precipitacion_mensual_estaciones.to_csv(
    f'{ruta_resultados}/precipitacion_mensual_estaciones.csv',
    sep=';',
    encoding='utf-8-sig'
)

resultado_mes_lluvioso.to_csv(
    f'{ruta_resultados}/mes_mas_lluvioso_estaciones.csv',
    sep=';',
    encoding='utf-8-sig'
)

print("Resultados del Ejercicio 2 exportados correctamente")
```
